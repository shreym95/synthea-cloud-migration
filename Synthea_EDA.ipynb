```python
import pandas as pd
import os
from google.colab import drive

# drive.mount('/content/drive')
```

```python
# Reading csv files from input

input_folder = '/content/drive/My Drive/Synthea_Raw_Data' # Replace with your file path

files = os.listdir(input_folder)
csvs = [csv for csv in files if csv.endswith('.csv')]

print(csvs)
```

```python
# Getting count of rows and columns for each file

try:
  for csv in csvs:
    file_path = os.path.join(input_folder, csv)
    df = pd.read_csv(file_path)
    print(csv.split(".")[0], len(df), len(df.columns), sep=",")
    # display(df.head())
except FileNotFoundError:
  print(f"Error: The file was not found at {file_path}")
except Exception as e:
  print(f"An error occurred: {e}")
```

```python
import pandas as pd

def check_data_quality(file_path):
    """
    Reads a CSV file and identifies common data quality issues.

    Args:
        file_path (str): The path to the CSV file.

    Returns:
        dict: A dictionary containing information about data quality issues.
    """
    try:
        df = pd.read_csv(file_path)
        print(f"Successfully read file: {file_path}")

        quality_report = {}

        # Check for missing values
        missing_values = df.isnull().sum()
        missing_values = missing_values[missing_values > 0]
        if not missing_values.empty:
            quality_report['missing_values'] = missing_values.to_dict()
            print("\n--- Missing Values ---")
            print(missing_values)
        else:
            quality_report['missing_values'] = "No missing values found."
            print("\n--- Missing Values ---")
            print("No missing values found.")

        # Check for duplicate rows
        duplicate_rows = df.duplicated().sum()
        if duplicate_rows > 0:
            quality_report['duplicate_rows'] = duplicate_rows
            print(f"\n--- Duplicate Rows ---")
            print(f"Found {duplicate_rows} duplicate rows.")
        else:
            quality_report['duplicate_rows'] = "No duplicate rows found."
            print(f"\n--- Duplicate Rows ---")
            print("No duplicate rows found.")

        # Basic data type check (display info)
        print("\n--- Data Types ---")
        print(df.info())
        # You can add more sophisticated type checking here if needed

        # You can add more checks here, like:
        # - Checking for values outside expected ranges
        # - Checking for inconsistent formatting

        return quality_report

    except FileNotFoundError:
        print(f"Error: The file was not found at {file_path}")
        return {"error": f"File not found: {file_path}"}
    except Exception as e:
        print(f"An error occurred: {e}")
        return {"error": str(e)}

csvs = ['organizations.csv', 'procedures.csv', 'patients.csv','providers.csv','encounters.csv','conditions.csv']
for csv_file in csvs:
    file_path = os.path.join(input_folder, csv_file)
    print(f"\n--- Checking file: {csv_file} ---")
    report = check_data_quality(file_path)
    # print("\n--- Quality Report ---")
    # print(report)
```

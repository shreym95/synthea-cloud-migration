# Extracting data from Addresses CSV

```python
import sys
import os

# Add the directory containing config.py to sys.path if needed
sys.path.append(os.path.abspath(''))

import config as cfg
```

```python

display(
    spark.sql(f"""
        SELECT * FROM 
        read_files('{cfg.VOLUME_PATH}/landing/operational_data/addresses',
        format => 'csv',
        delimiter => '\t',
        header => true)
    """
    )
)
```

```python
spark.sql(f"""
          CREATE OR REPLACE VIEW {cfg.CATALOG}.{cfg.BRONZE}.v_addresses
          AS        
          SELECT * FROM 
            read_files('{cfg.VOLUME_PATH}/landing/operational_data/addresses',
            format => 'csv',
            delimiter => '\t',
            header => true)
""")

display(spark.sql(f"""SELECT * FROM {cfg.CATALOG}.{cfg.BRONZE}.v_addresses"""))
```

# GizmoBox Data Lakehouse - Environment Setup


### 1. Creating External Location for the container in ADLS

```python
%sql

-- Setting up external location in the catalog
CREATE EXTERNAL LOCATION IF NOT EXISTS gizmobox_ext
URL 'abfss://gizmobox@dbassstorage.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `dbass-cred-sc`)
COMMENT 'External location for GizmoBox';

```

### 2. Access the gizmobox container

```python
# Wrapper function to connect to data lake and list files
def list_files(data_lake, container, directory=None):
    # Connecting to Data Lake
    connection = f'abfss://{container}@{data_lake}.dfs.core.windows.net'
    if directory:
        connection += f'/{directory}'

    display(
        dbutils.fs.ls(connection)
    )

list_files(
    data_lake = 'dbassstorage',
    container = 'gizmobox',
    directory = 'landing'
)


```

### 3. Create the catalog - Gizmobox AND respective schemas

```python
%sql

SHOW CATALOGS;
```

```python
%sql
CREATE CATALOG IF NOT EXISTS gizmobox
MANAGED LOCATION 'abfss://gizmobox@dbassstorage.dfs.core.windows.net'
COMMENT 'Catalog for GizmoBox project';

USE CATALOG gizmobox;
```

```python
%sql
-- Creating schemas

SHOW SCHEMAS;

-- Landing schema
CREATE SCHEMA IF NOT EXISTS landing
MANAGED LOCATION 'abfss://gizmobox@dbassstorage.dfs.core.windows.net';

-- First layer bronze schema
CREATE SCHEMA IF NOT EXISTS bronze
MANAGED LOCATION 'abfss://gizmobox@dbassstorage.dfs.core.windows.net';

-- Second layer silver schema
CREATE SCHEMA IF NOT EXISTS silver
MANAGED LOCATION 'abfss://gizmobox@dbassstorage.dfs.core.windows.net';

-- Third layer gold schema
CREATE SCHEMA IF NOT EXISTS gold
MANAGED LOCATION 'abfss://gizmobox@dbassstorage.dfs.core.windows.net';
```

```python
%sql
-- Check if schemas are created
SHOW SCHEMAS;
```
